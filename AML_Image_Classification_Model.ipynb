# **Import the Important lib**

```python
# tensorflow is an opensource framework and is used to help us in creating our own Neural network ML models from scratch

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# we will use keras, which could be considered as a library inside tensorflow, to get our dataset

fashion_mnist = tf.keras.datasets.fashion_mnist

# we are directly splitting the data into train and test sets when we use the load_data()

(train_images,train_labels),(test_images,test_labels) = fashion_mnist.load_data()
```

```python
# shape (number of imgs , size of imgs )

train_images.shape
```

```python
# to know the unique labels we have (the categories)

np.unique(train_labels)
```

```python
for i in range(9):
  # 3 rows , 3 columns, 0 is the starting point of the item in our row then i+1 for increasing by 1
  plt.subplot(330+i+1)
  plt.imshow(train_images[i],cmap = 'gray')
```

```python
# Normalization for the values
# scaling the pixel values so they lie in a smaller range to make the training process easier ( 0 - 1)
# by dividing on the maximum value 255.0
train_images = train_images/255.0
test_images = test_images/255.0
```

```python
# model creation
# we are using keras here because it foucses on deep learning and provides utilites to help in the development process
# any model will have an input layer, a possible amout of hidden layers, and an output layer
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28,28)), # Input Layer flatten means we are getting the total amount of pixels given as input to the model
    tf.keras.layers.Dense(128,activation="relu"),
    tf.keras.layers.Dense(10,activation="softmax") # Output layer, softmax function returns a probability between 0 and 1
    ])
```

```python
# Summry of the model

model.summary()
```

```python
# deciding on how to improve the model's learning process through the optimizer that adjusts the weight (Adam).
# the loss function technique (SparseCategoricalCrossentropy).

# This is a specific type of loss function used for multi-class classification problems
# where the target labels are integers so we are calculating the loss between the predicitons and the true labels
# by using from_logits=True, it ensures that the training process is more stable and accurate.

# and lastly evaluating the model performance (accuracy),

model.compile(optimizer="adam",loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),metrics=["accuracy"])
```

```python
# training the model
model.fit(train_images,train_labels)
```

```python
# extra not necessary step just to ensure the output will always be a probability

probability_model = tf.keras.Sequential([model,
                                         tf.keras.layers.Softmax()])
```

```python
# prediction

predictions = probability_model.predict(test_images)

#  what is the presentg for each category the sum is 1 or 100%
predictions
```

```python
# There are 10 output labels for the Fashion MNIST dataset

labels = ['t_shirt', 'trouser', 'pullover', 'dress', 'coat',
          'sandal', 'shirt', 'sneaker', 'bag', 'ankle_boots']
```

```python
# selecting the first image in the test to predict its label

predictions = probability_model.predict(test_images[:1])

print(predictions)
```

```python
# get the index of the highest probability from the preictions
# then decide on the label based on the index

label = labels[np.argmax(predictions)]

print(label)
```
