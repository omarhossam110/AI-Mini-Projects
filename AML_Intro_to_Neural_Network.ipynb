# What is a neural network?



*   The brain is made up of a lot of neurons this is how we can process any information and then take a decision
* First we receive input through the dendrites, goes through the cell body which sums up the different inputs receieved. Then it fires the axon which then could pass this received signal to another set of dendrites of another neuron. The space between the ending of the first neuron and the dendrites of the second neuron is called synapse.
* For example your finger touches a hot cup, the dendrites of the neuron recieve the input sensation of hurt, passes the information to the cell body and the axon and consequently to another cell until the brain process that, and as an output we need to remove our finger away from the hot cup.

**Artificial Neural Networks**


*   The artificial neurons tries to mimic the real ones in the human brian where they also take as input information for example a certain set of features, passes them to other neurons to try to understand them and recognize certain patterns and in the end come up with an output about what were these features making such as a cat for example
*   Let's try to simulate how we can take a descision let's say we are deciding should we go swimming at the beach or not so we have 3 main factors to decide on. Are there any sharks in the water?. Is there any jellyfish? And lastly, is the beach crowded?
* Now of course not all three factors have the same importance (weights) so we give each factor it's weight of importance from 1 to 5. Sharks takes **3**, JellyFish takes **2** and crowdedness takes **4**.
  1. Sharks: Weight = 3, Input = (No) 0 → Contribution = 0
  2. Jellyfish: Weight = 2, Input = (No) 0 → Contribution = 0
  3. Crowdedness: Weight = 4, Input = (No) 1 → Contribution = 4
  * Lastly we could decide on the decision by using an activation function that acts like having a threshold, if the weighted sum passes a certain threshold for example in our case 5 then we go swimming otherwise we don't. In the case of neurons in the hidden layers the activiation function decided if the information from this neuron will be passed to next one or not.

```python
# important libs

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
```

```python
# read the data
df = pd.read_csv("/content/dynamic_api_call_sequence_per_malware_100_0_306.csv")
```

```python
# Head
df.head()
```

```python
# prapration drop hash

df.drop("hash",axis=1,inplace=True)
```

```python
# drop any null values

df = df.dropna(how="any")
df.shape
```

# **Implementation**

```python
X = df.drop("malware",axis=1)
Y = df["malware"]
```

```python
# split the data

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
# import the model
from sklearn.neural_network import MLPClassifier

# Implement the model
mlp = MLPClassifier(hidden_layer_sizes=(100,100,100),max_iter=1000)

# Train
mlp.fit(x_train,y_train)

# predict
predictions = mlp.predict(x_test)

# accuracy
print(accuracy_score(y_test,predictions))
```
